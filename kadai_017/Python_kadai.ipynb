```python
class Human:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def check_adult(self):
    if 20 <= self.age:
      print(f"{self.name}は大人")
    else:
      print(f"{self.name}は大人じゃない")

member1 = Human("小林",30)
member2 = Human("田中",19)
member3 = Human("佐藤",20)
members = [member1,member2,member3]
# members = [Human("小林",30),Human("田中",19),Human("佐藤",20)]

for member in members:
  member.check_adult()
```
